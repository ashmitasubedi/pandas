```python
import pandas as pd

df = pd.read_csv("3) Sentiment dataset.csv")

print(df.describe())
```

```python
# 1. View basic information
print(df.info())
print(df.head())
```

```python
# 2. Remove unnecessary index column
if 'Unnamed: 0' in df.columns:
    df.drop('Unnamed: 0', axis=1, inplace=True)
```

```python
# 3. Check missing values
print(df.isnull().sum())
```

```python
# 1. View basic information
print(df.info())
print(df.head())
```

```python
# Remove unwanted index column
df.drop("Unnamed: 0.1", axis=1, inplace=True)
```

```python
# Remove leading/trailing spaces from all text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python

# Convert timestamp to datetime
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# Standardize text columns
df["Sentiment"] = df["Sentiment"].str.capitalize()
df["Platform"] = df["Platform"].str.capitalize()
df["Country"] = df["Country"].str.title()
```

```python
print(df.info())
print(df.head())
```

```python
# Save cleaned file
df.to_csv("cleaned_social_media_sentiment.csv", index=False)
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nUnique Sentiments:")
print(df["Sentiment"].unique())

print("\nUnique Platforms:")
print(df["Platform"].unique())

print("\nUnique Countries:")
print(df["Country"].unique())
```

```python
# Sentiment counts
print(df["Sentiment"].value_counts())

# Platform counts
print(df["Platform"].value_counts())

# Top countries
print(df["Country"].value_counts().head(10))

# Average likes by sentiment
print(df.groupby("Sentiment")["Likes"].mean())

# Average retweets by sentiment
print(df.groupby("Sentiment")["Retweets"].mean())
```
