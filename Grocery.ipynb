```python
file_path = r"C:\Users\Dell\Desktop\pandas\GroceryData.xlsx"

df = pd.read_excel(file_path)   # reads first sheet by default



```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df = df.dropna(how="all")   # remove fully empty rows
df = df.dropna(axis=1, how="all")  # remove fully empty columns
```

```python
df.isna().sum()
```

```python
df['item_weight'] = df.groupby('item_identifier')['item_weight']\
                      .transform(lambda x: x.fillna(x.mean()))
```

```python
df.isna().sum()
```

```python
df['item_weight'].describe()
```

```python
Q1 = df['item_weight'].quantile(0.25)
Q3 = df['item_weight'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['item_weight'] < lower) | (df['item_weight'] > upper)]
```

```python
df['item_weight'] = df['item_weight'].round(2)
```

```python
df['item_fat_content'] = (
    df['item_fat_content']
    .str.strip()
    .str.lower()
    .replace({
        'lf': 'low fat',
        'low fat': 'low fat',
        'reg': 'regular',
        'regular': 'regular'
    })
)
```

```python
df['item_identifier'] = df['item_identifier'].str.strip().str.upper()
```

```python
df['item_type'] = df['item_type'].str.strip().str.title()
```

```python
# Step 2 fallback
df['item_weight'] = df['item_weight'].fillna(df['item_weight'].median())
```

```python
df['item_visibility'] = df['item_visibility'].round(6)
df['item_weight'] = df['item_weight'].round(2)
df['sales'] = df['sales'].round(2)
```

```python
df['item_visibility'] = df['item_visibility'].replace(
    0, df['item_visibility'].median()
)
```
